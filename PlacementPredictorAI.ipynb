```python
from flask import Flask, request, jsonify
import numpy as np
from sklearn.linear_model import LogisticRegression
from sklearn.preprocessing import LabelEncoder
import joblib

# 1. Train the model
X = np.array([[1, 0], [2, 1], [3, 0], [4, 1]])
y = np.array(['Placed', 'Not Placed', 'Placed', 'Not Placed'])

# Encode labels
label_enc = LabelEncoder()
y_encoded = label_enc.fit_transform(y)

# Train model
model = LogisticRegression()
model.fit(X, y_encoded)

# Save model and encoder
joblib.dump(model, 'placement_model.pkl')
joblib.dump(label_enc, 'label_encoder.pkl')

# 2. Set up the Flask app
app = Flask(__name__)

@app.route('/')
def home():
    return "Welcome to the Placement Predictor API!"

@app.route('/predict', methods=['POST'])
def predict():
    data = request.get_json()
    features = np.array(data['features']).reshape(1, -1)
    prediction = model.predict(features)
    label = label_enc.inverse_transform(prediction)
    return jsonify({'prediction': label[0]})

if __name__ == '__main__':
    app.run(debug=True, port=5002)  # Run Flask on port 5002
```

```python
from flask import Flask, request, jsonify, render_template_string
import joblib
import numpy as np
from pyngrok import ngrok

# Load saved model and label encoder
model = joblib.load('placement_model.pkl')
label_enc = joblib.load('label_encoder.pkl')

app = Flask(__name__)

@app.route('/')
def home():
    html_content = """
    <!DOCTYPE html>
    <html lang="en">
    <head>
        <meta charset="UTF-8">
        <meta name="viewport" content="width=device-width, initial-scale=1.0">
        <title>Placement Predictor</title>
        <style>
            body {
                font-family: Arial, sans-serif;
                text-align: center;
                margin-top: 50px;
                background: #f5f5f5;
            }

            .container {
                background: white;
                padding: 20px;
                display: inline-block;
                border-radius: 10px;
                box-shadow: 0 0 10px gray;
            }

            input {
                margin: 10px;
                padding: 5px;
                width: 100px;
            }

            button {
                padding: 8px 16px;
                background-color: #4CAF50;
                color: white;
                border: none;
                cursor: pointer;
                font-weight: bold;
                border-radius: 5px;
            }

            button:hover {
                background-color: #45a049;
            }
        </style>
    </head>
    <body>
        <div class="container">
            <h1>Placement Predictor</h1>
            <label>Enter Feature 1: <input type="number" id="feature1"></label><br>
            <label>Enter Feature 2: <input type="number" id="feature2"></label><br>
            <button onclick="getPrediction()">Predict</button>
            <h3 id="result"></h3>
        </div>

        <script>
            function getPrediction() {
                const feature1 = parseFloat(document.getElementById('feature1').value);
                const feature2 = parseFloat(document.getElementById('feature2').value);

                fetch('/predict', {
                    method: 'POST',
                    headers: {
                        'Content-Type': 'application/json',
                    },
                    body: JSON.stringify({
                        features: [feature1, feature2]
                    })
                })
                .then(response => response.json())
                .then(data => {
                    document.getElementById('result').innerText = "Prediction: " + data.prediction;
                })
                .catch(error => {
                    document.getElementById('result').innerText = "Error: " + error;
                });
            }
        </script>
    </body>
    </html>
    """
    return render_template_string(html_content)

@app.route('/predict', methods=['POST'])
def predict():
    try:
        data = request.get_json()
        features = data.get('features')

        # Check if features are provided and are not empty
        if not features or len(features) != 2:
            return jsonify({'error': 'Invalid input. Please provide exactly two features.'}), 400

        # Convert the features to a numpy array and make a prediction
        features = np.array(features).reshape(1, -1)
        prediction = model.predict(features)
        label = label_enc.inverse_transform(prediction)

        return jsonify({'prediction': label[0]})

    except Exception as e:
        return jsonify({'error': str(e)}), 500

```
