```python
if mydata.isnull().sum().any():
    print("\nMissing values detected. Please clean the dataset before proceeding.")
else:
    print("\nNo missing values detected.")
```

```python
# Assuming the dataset has features starting from the second column
features = mydata.iloc[:, 1:]  # Exclude the first column if it's an identifier
print("\nSelected Features for Similarity Calculation:")
print(features.columns)
```

```python
scaler = StandardScaler()
scaled_features = scaler.fit_transform(features)

# Display the standardized features (optional)
print("\nStandardized Features (First 5 Rows):")
print(scaled_features[:5])
```

```python
similarity_matrix = cosine_similarity(scaled_features)
print("\nCosine Similarity Matrix Calculated.")
```

```python
def recommend_wines(wine_id, num_recommendations=5):
    """
    Recommends wines similar to the given wine_id based on cosine similarity.
    
    Parameters:
        wine_id (int): Index of the wine in the dataset.
        num_recommendations (int): Number of recommendations to return.

    Returns:
        DataFrame: Recommended wines with similarity scores.
    """
    # Get similarity scores for the given wine
    similarity_scores = list(enumerate(similarity_matrix[wine_id]))
    
    # Sort wines by similarity scores (descending order)
    sorted_wines = sorted(similarity_scores, key=lambda x: x[1], reverse=True)
    
    # Exclude the input wine itself
    sorted_wines = sorted_wines[1:num_recommendations + 1]
    
    # Get the recommended wine indices and scores
    recommended_indices = [i[0] for i in sorted_wines]
    similarity_scores = [i[1] for i in sorted_wines]
    
    # Return recommended wines with their similarity scores
    recommended_wines = mydata.iloc[recommended_indices].copy()
    recommended_wines['Similarity_Score'] = similarity_scores
    return recommended_wines
```

```python
# Example: Get recommendations for the first wine in the dataset
wine_id = 0  # Change this to the desired wine index
recommendations = recommend_wines(wine_id, num_recommendations=5)

print(f"\nRecommended wines for Wine ID {wine_id}:")
print(recommendations)
```
